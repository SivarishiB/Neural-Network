## Training the data

```python
from sklearn.model_selection import train_test_split, GridSearchCV
```

```python
X = df.drop("Y", axis=1)
y = df["Y"]
```

```python
X.shape
```

```python
 y.shape
```

```python
y = y.map({
    "R": 0,
    "M": 1
})

print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## Data Processing

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
```

```python
#Data Processing
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler()),
    ("ann", MLPClassifier(
        hidden_layer_sizes=(10,),
        activation="relu",
        learning_rate_init=0.001,
        max_iter=1000,
        random_state=42
    ))
])
```

## ANN Model

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)

print(y_pred)
```

## Evaluating the ANN Model

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
precision = precision_score(y_test, y_pred)

print("Precision:", precision)
```

```python
recall = recall_score(y_test, y_pred)

print("Recall:", recall)
```

```python
f1 = f1_score(y_test, y_pred)

print("F1 Score:", f1)
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

## Data Visualisation

```python
plt.imshow(cm)

plt.title("Confusion Matrix - Basic ANN")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.xticks([0, 1], ["Rock", "Mine"])
plt.yticks([0, 1], ["Rock", "Mine"])

plt.colorbar()

plt.show()
```

## Define Hyperparameters

```python
parameter_grid = {
    "ann__hidden_layer_sizes": [
        (5,),
        (10,),
        (20,),
        (10, 5),
        (20, 10)
    ],

    "ann__activation": [
        "relu",
        "tanh"
    ],

    "ann__learning_rate_init": [
        0.001,
        0.01
    ]
}
```

```python
grid_search = GridSearchCV(
    pipeline,
    parameter_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)
```

## Tuned ANN

```python
best_model = grid_search.best_estimator_
```

```python
tuned_pred = best_model.predict(X_test)
```

```python
tuned_accuracy = accuracy_score(y_test, tuned_pred)
tuned_precision = precision_score(y_test, tuned_pred)
tuned_recall = recall_score(y_test, tuned_pred)
tuned_f1 = f1_score(y_test, tuned_pred)

print("Accuracy :", tuned_accuracy)
print("Precision:", tuned_precision)
print("Recall   :", tuned_recall)
print("F1 Score :", tuned_f1)
```

## Comparing the Models

```python
results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],

    "Basic ANN": [
        accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred),
        recall_score(y_test, y_pred),
        f1_score(y_test, y_pred)
    ],

    "Tuned ANN": [
        tuned_accuracy,
        tuned_precision,
        tuned_recall,
        tuned_f1
    ]
})

results
```

## Visualizing Model Comparison

```python
results.set_index("Metric").plot(kind="bar")

plt.title("Basic ANN vs Tuned ANN")
plt.ylabel("Score")
plt.ylim(0, 1)

plt.xticks(rotation=0)
plt.show()
```

## Conclusion

The ANN successfully classified sonar signals into Mine and Rock categories. Data preprocessing, feature scaling and a train/test split were performed before model training. A basic neural network was then improved using systematic hyperparameter tuning.

For the supplied dataset and selected random split, the tuned ANN achieved approximately 88.10% accuracy and 88.89% F1-score on the test set. The higher recall for the Mine class means the tuned model correctly identified more of the actual Mine samples in this particular test set.
